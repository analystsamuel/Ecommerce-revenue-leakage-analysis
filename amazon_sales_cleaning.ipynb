```python
for col in ["Courier Status", "currency", "ship-city", "ship-state", "ship-country", "fulfilled-by"]:
    df[col] = df[col].fillna("Unknown")
df["Amount"] = df["Amount"].fillna(0)
```

```python
state_fixes = {"Rajshthan": "Rajasthan","Rajsthan": "Rajasthan","Rj": "Rajasthan","Nl": "Nagaland","New Delhi": "Delhi",
               "Punjab/Mohali/Zirakpur": "Punjab","Pb": "Punjab","Orissa": "Odisha","Pondicherry": "Puducherry","Ar": "Arunachal Pradesh",
               "Apo": "Unknown"}
df["ship_state"] = df["ship_state"].replace(state_fixes)
```
